```python
import pandas as pd
import numpy as np
import pickle

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
```

```python
X_train = pd.read_csv("../data/processed/X_train.csv")
X_val = pd.read_csv("../data/processed/X_val.csv")

y_train = pd.read_csv("../data/processed/y_train.csv")
y_val = pd.read_csv("../data/processed/y_val.csv")

print("Train shape:", X_train.shape)
print("Validation shape:", X_val.shape)
```

```python
y_train = y_train.values.ravel()
y_val = y_val.values.ravel()

print("Target prepared")
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

print("Model training completed")
```

```python
y_pred_prob = model.predict_proba(X_val)[:,1]

print("Predictions generated")
```

```python
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score
import json

roc_auc = roc_auc_score(y_val, y_pred_prob)
precision = precision_score(y_val, model.predict(X_val))
recall = recall_score(y_val, model.predict(X_val))
f1 = f1_score(y_val, model.predict(X_val))

print("Baseline ROC-AUC:", round(roc_auc,4))
print("Baseline Precision:", round(precision,4))
print("Baseline Recall:", round(recall,4))
print("Baseline F1:", round(f1,4))

baseline_metrics = {
    "Model": "Logistic Regression",
    "ROC_AUC": roc_auc,
    "Precision": precision,
    "Recall": recall,
    "F1": f1,
}

with open("../data/processed/baseline_metrics.json", "w") as f:
    json.dump(baseline_metrics, f, indent=4)

print("baseline_metrics.json saved")
```
